# Jiminy Engine — Example Usage

This notebook demonstrates how to use the Jiminy Engine to perform normative reasoning and argumentation on a scenario.

```python
import sys
import os

# Add the project root to the path so we can import jiminy
sys.path.insert(0, os.path.abspath('..'))

from jiminy.engine import Jiminy
from jiminy.yaml_loader import load_scenario
```

## Loading a Scenario

We will load the `agrobot.yaml` scenario, which models a decision-making process for an autonomous agricultural robot.

```python
scenario_path = "../scenarios/agrobot.yaml"
(
    possible_facts,
    norms,
    contrariness,
    priorities,
    context_desc,
    norm_desc,
    contrariness_desc,
    priority_desc
) = load_scenario(scenario_path)

print(f"Loaded scenario from {scenario_path}")
print(f"Context facts: {list(possible_facts)}")
```

## Initializing Jiminy

We create a `Jiminy` instance with the loaded scenario data.

```python
jim = Jiminy(
    norms,
    contrariness,
    priorities,
    context_desc,
    norm_desc,
    contrariness_desc,
    priority_desc
)
```

## Reasoning and Explanation

We generate arguments from the context facts and compute the extension using the `jiminy` semantics. Finally, we print the narrative explanation.

```python
print("Generating arguments...")
args = jim.generate_arguments(possible_facts)

print("Computing extension (semantics='jiminy')...")
accepted, rejected = jim.compute_extension(args, semantics="priority")

print("\n--- Explanation ---\n")
print(jim.explain(accepted, rejected, possible_facts))
```
